```python
import matplotlib.pyplot as plt

# One filtered table, copied so pandas doesn't warn
per = session.laps.pick_drivers('PER').copy()   # or: session.laps[session.laps['Driver'] == 'PER'].copy()

# Lap time in seconds
per['LapTimeSec'] = per['LapTime'].dt.total_seconds()

# Colours for each compound
colors = {
    'SOFT': 'red',
    'MEDIUM': 'gold',
    'HARD': 'grey',        # white would be invisible on a white background
    'INTERMEDIATE': 'green',
    'WET': 'blue',
}

plt.figure(figsize=(12, 5))

for compound, group in per.groupby('Compound'):
    plt.scatter(
        group['LapNumber'],
        group['LapTimeSec'],
        color=colors.get(compound, 'black'),   # black for any unexpected value
        label=compound,
        edgecolor='black',
        s=40,
    )

# Faint line through all laps so the race flow is visible
plt.plot(per['LapNumber'], per['LapTimeSec'], color='lightgrey', zorder=0)

# Stop a few slow laps from squashing the chart
fastest = per['LapTimeSec'].min()
plt.ylim(fastest - 1, fastest + 10)

plt.xlabel('Lap number')
plt.ylabel('Lap time (seconds)')
plt.title('Perez lap times by tyre compound')
plt.legend()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 9), sharex=True)

for ax in (ax1, ax2):
    for compound, group in per.groupby('Compound'):
        ax.scatter(
            group['LapNumber'],
            group['LapTimeSec'],
            color=colors.get(compound, 'black'),
            label=compound,
            edgecolor='black',
            s=40,
        )
    ax.plot(per['LapNumber'], per['LapTimeSec'], color='lightgrey', zorder=0)
    ax.set_ylabel('Lap time (seconds)')

# Top: everything, including spikes
ax1.set_title('Perez lap times: full view')
ax1.legend()

# Bottom: zoomed on normal racing laps
fastest = per['LapTimeSec'].min()
ax2.set_ylim(fastest - 1, fastest + 10)
ax2.set_title('Zoomed on normal laps')
ax2.set_xlabel('Lap number')

plt.tight_layout()
plt.show()
```
